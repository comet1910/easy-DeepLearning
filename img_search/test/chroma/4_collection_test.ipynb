```python
import  chromadb
```

```python
# 创建客户端
client = chromadb.PersistentClient(path='./chroma_db')
```

```python
client.list_collections()
```

## 创建集合

```python
# 1. 直接创建
collection1 = client.create_collection('collection1')
# 2. 获取或创建
collection2 = client.get_or_create_collection('collection2')
```

## 查看集合

```python
# 1. 列出所有集合
client.list_collections()
```

```python
# 2. 获取某个特定集合
collection2_cpy = client.get_collection('collection2')
```

```python
print(collection2)
print(collection2_cpy)
```

```python
print(collection2.id)
print(collection2_cpy.id)
```

## 向集合中添加数据

```python
collection2.add(
    ids=["1", "2", "3"],
    documents=[
        "This is the first document",
        "This is the second document",
        "This is the third document",
    ]
)
```

```python
collection2.peek()
```

```python
collection2_cpy.peek()
```

```python
collection2_cpy.add(
    ids=["4", "5", "6"],
    documents=[
        "This is the fourth document",
        "This is the fifth document",
        "This is the sixth document",
    ]
)
```

```python
collection2.peek()
```

```python
collection2_cpy.peek()
```

## 删除集合

```python
client.delete_collection('collection1')
```

```python
client.list_collections()
```

## 便捷方法

```python
# 1. 返回前几条数据
collection2.peek(limit=4)
```

```python
# 2. 返回数据条数
collection2_cpy.count()
```
