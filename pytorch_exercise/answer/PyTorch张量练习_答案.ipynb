### 2.4.2 Tensor 与 ndarray 转换

**练习 16**：`Tensor.numpy()` 与共享内存
1. 创建 `tensor1 = torch.rand(3, 2)`
2. 用 `.numpy()` 转为 ndarray，分别打印
3. 修改 `tensor1[:, 0] = 4`，再次打印 tensor 与 ndarray，观察 ndarray 是否同步变化

```python
# 练习 16：Tensor.numpy() 共享内存

tensor1 = torch.rand(3, 2)
numpy_array = tensor1.numpy()
print(tensor1)
print(numpy_array)

tensor1[:, 0] = 4
print(tensor1)
print(numpy_array)  # 与 tensor 共享内存，同步变化
```

**练习 17**：使用 `.numpy().copy()` 避免共享内存
1. 将 tensor 转为 ndarray 并 `copy()`
2. 修改原 tensor 的值，验证 copy 出来的 ndarray 不受影响

```python
# 练习 17：使用 copy() 避免共享内存

tensor1 = torch.rand(3, 2)
numpy_array = tensor1.numpy().copy()
tensor1[:, 0] = -1
print(tensor1)
print(numpy_array)  # 不受 tensor 修改影响
```

**练习 18**：`torch.from_numpy()` 与共享内存
1. 创建 numpy 数组 `np.random.randn(3)`
2. 用 `torch.from_numpy` 转为 tensor
3. 修改 numpy 数组第一个元素为 `100`，观察 tensor 是否同步变化

```python
# 练习 18：torch.from_numpy() 共享内存

numpy_array = np.random.randn(3)
tensor1 = torch.from_numpy(numpy_array)
print(numpy_array)
print(tensor1)

numpy_array[0] = 100
print(numpy_array)
print(tensor1)  # 与 numpy 共享内存，同步变化
```

**练习 19**：`torch.tensor(ndarray)` 不共享内存
1. 创建 numpy 数组并转为 tensor
2. 修改 numpy 数组，验证 tensor 保持不变

```python
# 练习 19：torch.tensor(ndarray) 不共享内存

numpy_array = np.random.randn(3)
tensor1 = torch.tensor(numpy_array)
print(numpy_array)
print(tensor1)

numpy_array[0] = 100
print(numpy_array)
print(tensor1)  # 不共享内存，保持不变
```

## 2.7 张量索引操作

**练习 36**：布尔索引进阶
对 `(3, 5, 4)` 张量，取第 1 维第 1 个元素大于 5 的位置，返回形状为 `(dim0, dim2)` 的结果（提示：需先用 `permute` 调整维度顺序再索引）。

```python
# 练习 36：布尔索引进阶

tensor1 = torch.randint(1, 9, (3, 5, 4))

mask = tensor1[:, 1, :] > 5
print(mask, mask.shape)

tensor2 = tensor1.permute(0, 2, 1)   # (dim0, dim2, dim1)
print(tensor2[mask], tensor2[mask].shape)
```

1. 创建 `(3, 5, 4)` 张量，用 `is_contiguous()` 判断是否内存连续，用 `view(-1, 10)` 调整形状
2. 对张量做转置 `.T`，再次判断内存连续性，并说明为何需要先 `contiguous()` 才能 `view`

```python
# 练习 39：view 与内存连续性

tensor1 = torch.randint(1, 9, (3, 5, 4))
print(tensor1.is_contiguous())
print(tensor1.view(-1, 10).shape)

tensor1 = tensor1.T
print(tensor1.is_contiguous())
print(tensor1.contiguous().view(-1).shape)  # 转置后非连续，需先 contiguous
```

## 2.9 张量拼接操作

**练习 41**：`torch.cat` 拼接
1. 创建形状 `(2, 2, 5)` 与 `(2, 1, 5)` 的两个张量
2. 在第 1 维上拼接（`dim=1`），打印结果形状

思考：为什么可以在 dim=1 上拼接（除拼接维度外其它维度需相同）？

```python
# 练习 41：torch.cat 拼接

tensor1 = torch.randint(1, 9, (2, 2, 5))
tensor2 = torch.randint(1, 9, (2, 1, 5))
print(tensor1)
print(tensor2)

tensor3 = torch.cat([tensor1, tensor2], dim=1)
print(tensor3, tensor3.shape)  # (2, 3, 5)
```

**练习 42**：`torch.stack` 堆叠
1. 创建两个形状均为 `(3, 1, 5)` 的张量
2. 在第 2 维上堆叠（`dim=2`），打印结果形状

思考：`cat` 与 `stack` 的区别是什么？

```python
# 练习 42：torch.stack 堆叠

torch.manual_seed(42)
tensor1 = torch.randint(1, 9, (3, 1, 5))
tensor2 = torch.randint(1, 9, (3, 1, 5))
print(tensor1)
print(tensor2)

tensor3 = torch.stack([tensor1, tensor2], dim=2)
print(tensor3, tensor3.shape)  # (3, 1, 2, 5)
```
